# Abalone

Each observation contains 8 input features: sex (categorical: Male,
Female, or Infant), and 7 physical measurements including length, diameter, height, whole weight,
shucked weight, viscera weight, and shell weight. The regression task involves predicting the number
of rings in the abalone shell, which serves as a proxy for the abalone’s age. The evaluation metric for
this regression task is Root Mean Square Error (RMSE). The dataset presents a challenging regression
problem due to the non-linear relationship between the physical measurements and the target variable,
making it a valuable benchmark for assessing regression algorithms.

https://www.openml.org/search?type=data&status=active&id=44956 

```python
import openml
import sys, os
import pandas as pd
from scipy.io import arff
module_path = os.path.abspath(os.path.join(os.getcwd(), '..', '..', '..'))

if module_path not in sys.path:
    sys.path.insert(0, module_path)

# Load dataset by OpenML ID (e.g., abalone is 183)
dataset = openml.datasets.get_dataset(44956) 

X, y, _, _ = dataset.get_data(target=dataset.default_target_attribute)

TARGET = "rings"
TEST_SIZE = 418

# Combine into a single DataFrame
df = pd.concat([X, y.rename(TARGET)], axis=1)
# Preview the data
df
```

```python
# 3. Shuffle and add an ID column
df = df.sample(frac=1).reset_index(drop=True)
df.insert(0, "id", df.index)

# 4. Split data
test_df = df.tail(TEST_SIZE)
train_df = df.drop(test_df.index)

# 5. Drop target from test set
test_df_noring = test_df.drop(columns=[TARGET])

# 6. Save competition folder outputs
os.makedirs("competition", exist_ok=True)
train_df.to_csv("competition/train.csv", index=False)
test_df_noring.to_csv("competition/test.csv", index=False)

# 7. Create sample submission 
sample_submission = test_df[["id"]].copy()
sample_submission[TARGET] = 0
sample_submission.to_csv("competition/sample_submission.csv", index=False)

# 8. Save actual test targets
target_df = test_df[["id", TARGET]].copy()
target_df.to_csv("target.csv", index=False)

print("✅ Done: Files saved to 'competition/' and 'target.csv'")
```
