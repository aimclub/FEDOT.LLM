```python
from pathlib import Path
import sys

module_path = Path.cwd().parent.parent.absolute()

if str(module_path) not in sys.path:
    sys.path.insert(0, str(module_path))

from fedotllm.main import FedotAI
from fedotllm.handlers import JupyterOutput

dataset_path = Path.cwd() / "competition"
dataset_path.mkdir(parents=True, exist_ok=True)
```

```python
import shutil

output_path = Path.cwd() / 'output'
if output_path.exists():
    shutil.rmtree(output_path)
output_path.mkdir(parents=True, exist_ok=True)
```

```python
import openml

dataset = openml.datasets.get_dataset(42705) 

X, y, _, _ = dataset.get_data(target=dataset.default_target_attribute)
```

```python
import pandas as pd

data = pd.concat([X, y.rename("101")], axis=1)
```

```python
data
```

```python
from sklearn.model_selection import train_test_split

train_data, test_data = train_test_split(data, test_size=0.1, random_state=42, shuffle=False)
y_test = test_data['101']
test_data = test_data.drop(columns='101')

print(len(train_data))
print(len(test_data))

train_data.to_csv(dataset_path/'train.csv', index=False)
test_data.to_csv(dataset_path/'test.csv', index=False)
# y_test.to_csv(dataset_path/'sample_submission.csv')
```

```python
description = \
"""
Your Goal:
This dataset was utilized for a regression task. Good luck!
Evaluation
The evaluation metric for this regression task is Root Mean Square Error (RMSE)
Submission File
Objective is to predict the values in column "101" based on the preceding features.
"""
```

```python
fedot_ai = FedotAI(
        task_path=dataset_path,
        workspace=output_path,
        handlers=JupyterOutput().subscribe
    )

async for _ in fedot_ai.ask(message=description):
    continue
```
